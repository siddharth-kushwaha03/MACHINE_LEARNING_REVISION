```python
import pandas as pd 
import numpy as np 

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 28, 35, np.nan, 40, 27, 32],
    "Salary": [25000, 30000, 28000, np.nan, 40000, 45000, 35000, np.nan, 32000, 50000],
    "Experience": [1, 2, 2, 5, np.nan, 8, 4, 10, 3, np.nan],
    "Gender": ["Male", "Female", "Male", np.nan, "Female", "Male", "Female", "Male", "Female", "Male"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "IT", "HR", "Finance"],
    "Purchased": ["No", "Yes", "No", "Yes", "Yes", "Yes", "No", "Yes", "No", "Yes"]
})

df
```

##  🟢 Phase 3 — Set 2: Encoding + Scaling Practice

```python
# Q1. Missing values identify karo.

df.isnull().sum()
```

```python
#Q2. Numerical columns ke missing values ko median se fill karo.

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df["Experience"] = df["Experience"].fillna(df["Experience"].median())
```

```python
# Q3. Categorical column ke missing values ko mode se 
# fill karo.

df["Gender"]=df["Gender"].fillna(df["Gender"].mode()[0])
```

```python
# Q4. Binary categorical column ko 0/1 mein encode karo. 

df["Gender"]=df["Gender"].map({"Male":0,"Female":1})

df["Purchaesd"]=df["Purchased"].map({"No":0,"Yes":1})
```

```python
#Q5. Multi-category column par One-Hot Encoding lagao.

df=pd.get_dummies(df,columns=["Department"],dtype=int)
df.head()
```

```python
# Q6. X aur y separate karo.

X=df.drop("Purchased",axis=1)

y=df["Purchased"]
```

```python
# Q7. train_test_split() se 80:20 split karo.

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
# Q8. Numerical features par StandardScaler apply karo.

scaler=StandardScaler()

num_cols=["Age","Salary","Experience"]

X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
```

```python
# Q9. Verify karo ki training data scaled hai.

print(X_train[num_cols].mean())
print(X_test[num_cols].std())
```
