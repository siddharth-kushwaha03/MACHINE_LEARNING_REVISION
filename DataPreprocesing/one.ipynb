```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, 21, np.nan, 23, 24, 21, np.nan, 25],
    "Salary": [25000, 30000, np.nan, 35000, 40000, 30000, np.nan, 45000],
    "Experience": [1, 2, 1, np.nan, 4, 2, np.nan, 5],
    "City": ["Lucknow", "Delhi", "Lucknow", np.nan, "Delhi", "Delhi", "Lucknow", "Mumbai"],
    "Gender": ["Male", "Female", "Male", "Female", np.nan, "Female", "Male", "Female"],
    "Purchased": ["No", "Yes", "No", "Yes", "Yes", "No", "Yes", "Yes"]
})
print(df)
```

## 🟢 Level 1 — Missing Values

Q1-Check karo ki har column mein kitne missing values hain.

```python
print(df.isnull().sum())
```

Q2.Age ke missing values ko mean se fill karo.

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Q3-Salary ke missing values ko median se fill karo.

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df
```

```python
# 4.Experience ke missing values ko mean se fill karo 

df["Experience"] = df["Experience"].fillna(df["Experience"]
                                           .mean()
)
```

```python
# 5.City ke missing values ko mode se fill karo.

df["City"]=df["City"].fillna(df["City"].mode())
df["City"]
```

```python
#  6 Gender ke missing values ko mode se fill karo.

df["Gender"]=df["Gender"].fillna(df["Gender"].mode()[0])
df["Gender"]
```

```python
# 7-Confirm karo ki ab DataFrame mein koi missing value nahi hai.

df.isnull().sum()
```

```python
# 8-Gender ko numerical form mein convert karo:

df["Gender"]=df["Gender"].map({"NO":0,"Yes":1})
df["Gender"]
```
