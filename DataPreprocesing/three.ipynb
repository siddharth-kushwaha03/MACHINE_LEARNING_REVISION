```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer


df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 28, 35, 24, np.nan],
    "Salary": [25000, 30000, 35000, np.nan, 45000, 50000, np.nan, 40000],
    "City": ["Lucknow", "Delhi", "Mumbai", "Delhi", np.nan, "Mumbai", "Lucknow", "Delhi"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", np.nan, "Female", "Male"],
    "Purchased": [0, 1, 0, 1, 1, 1, 0, 1]
})

print(df)
```

```python
#Q1.Features aur target separate karo.

X=df.drop("Purchased",axis=1)
y=df["Purchased"]
```

```python
# Q2.Numerical columns identify karo.
# Aur categorical columns identify karo.

num_cols=["Age","Salary"]

cat_cols=["City","Gender"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
# 3.Numerical columns ke liye:

# Missing values → Median
# Scaling → StandardScaler

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
# 4.Categorical columns ke liye:

# Missing values → Most Frequent
# Encoding → OneHotEncoder

cat_pipeline= Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
     ("encoder",OneHotEncoder(handle_unknown="ignore"))
])

```

```python
# 🟢 Q5 — ColumnTransformer
# Ab numerical aur categorical pipelines ko combine karenge.

from sklearn.compose import ColumnTransformer

preprocessor=ColumnTransformer([("num",num_pipeline,num_cols),
                                ("cat",cat_pipeline,cat_cols)])
     
```
