```python

# # Q4. Bagging kya hai?

# Bagging mein multiple models ko training data ke alag-alag bootstrap samples par train 
# kiya jata hai. Phir unke 
# predictions combine hote hain.
```

```python
# # # Q5. Boosting kya hai?

# Boosting mein models sequentially train hote hain. Har naya model pehle models ki mistakes ko 
# improve karne ki koshish karta hai.
```

```python
# # # Q6. Bagging aur Boosting mein difference batao.


# Models independently train ho sakte hain

# Models sequentially train hote hain

# Variance kam karne mein useful

# Weak models ko sequentially improve karta hai

# Example: Random Forest	
# Example: AdaBoost
```

```python
# # Q7. Random Forest kya hai?

# Random Forest multiple Decision Trees ka ensemble hai.
# Classification mein trees ke votes se final class decide hoti hai.
```

```python
# # Q8. Random Forest mein multiple Decision Trees kaise kaam karte hain?

# Training Dataset
#        ↓
# Multiple Samples
#        ↓
#  ┌─────┼─────┐
# Tree 1 Tree 2 Tree 3
#  └─────┼─────┘
#        ↓
#   Majority Vote
#        ↓
#    Final Output
```
